# AI vs Human Text Detector Using Simple RNN

```python
#Import Libraries
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
#Small Dataset
texts = [
    "I went to market yesterday",
    "The weather was very nice today",
    "Artificial intelligence improves efficiency",
    "Technology transforms industries rapidly",
    "I played cricket with friends",
    "Machine learning automates tasks"
]

labels = [0, 0, 1, 1, 0, 1]   #0 - AI
                              #1 - Human
```

```python
#Tokenization
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)

X = tokenizer.texts_to_sequences(texts)
X = pad_sequences(X)

y = np.array(labels)
```

```python
#Build RNN Model
model = Sequential([
    Embedding(100, 8),

    SimpleRNN(16),

    Dense(2, activation='softmax')
])
```

```python
#Train Model
model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.fit(X, y, epochs=200, verbose=0)
```

```python
#Test Model
test = ["AI is changing the world"]

seq = tokenizer.texts_to_sequences(test)
seq = pad_sequences(seq, maxlen=X.shape[1])

pred = np.argmax(model.predict(seq))

classes = ["Human", "AI"]

print("Prediction:", classes[pred])
```
