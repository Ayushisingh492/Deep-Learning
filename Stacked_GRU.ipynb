# Password Strength Predictor Using Stacked GRU

```python
#Import Libraries
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GRU, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
#Small Dataset
passwords = [
    "12345",
    "password",
    "ayushi123",
    "Hello@123",
    "Strong#99",
    "Ayushi@903"
]

labels = [0, 0, 0, 1, 2, 2] #{Classes:
                                     #0 = Weak
                                     #1 = Medium
                                     #2 = Strong}
```

```python
#Tokenization
tokenizer = Tokenizer(char_level=True)
tokenizer.fit_on_texts(passwords)

X = tokenizer.texts_to_sequences(passwords)
X = pad_sequences(X)

y = np.array(labels)
```

```python
#Build Stacked GRU Model
model = Sequential([
    Embedding(50, 8),

    GRU(16, return_sequences=True),

    GRU(16),

    Dense(3, activation='softmax')
])
```

```python
#Train Model
model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.fit(X, y, epochs=200, verbose=0)
```

```python
#Test Password
test = ["Admin@123"]

seq = tokenizer.texts_to_sequences(test)
seq = pad_sequences(seq, maxlen=X.shape[1])

pred = np.argmax(model.predict(seq))

classes = ["Weak", "Medium", "Strong"]

print("Password Strength:", classes[pred])
```
